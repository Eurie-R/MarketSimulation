# Market ABM sandbox: transaction tax and trader heterogeneity

**These results illustrate the model's mechanism, not the Philippine Stock Exchange.** The model is a stylized limit-order-book market with fundamentalist, chartist and noise traders; a seller-side tax filters out trades whose expected gain is below the tax.

Set the sliders and press **Run**. A run with T = 10,000 steps takes a few seconds.

```python
%matplotlib inline
from pathlib import Path

from marketabm.config import load_model_config
from marketabm.viz import compare_ui, live_ui, sandbox_ui

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
cfg_path = ROOT / "configs" / "calibrated.yaml"
if not cfg_path.exists():
    cfg_path = ROOT / "configs" / "baseline.yaml"
base = load_model_config(cfg_path)
print("base config:", cfg_path.name)
```

## 1. Single run

Shares are normalized to sum to 1 when you press Run. Output: price vs fundamental value (burn-in shaded), period returns, volume share by trader type, spread, volume, return distribution vs normal, ACFs and a stylized-facts card with pass/fail marks against the validation targets.

```python
sandbox_ui(base)
```

## 2. Compare two tax rates

Runs the **same seed** at two tax rates. Because all random numbers come from fixed per-purpose streams (common random numbers), the fundamental path, agents, activation order and noise are identical, so any difference is caused by the tax. Defaults are the old (0.6%) and new (0.1%) Philippine STT rates.

```python
compare_ui(base)
```

## 3. Live market

Watch the market move step by step. This drives the same simulator as `run()` (same seed, same path), stepping it in the background and redrawing after every frame.

- **Model settings** (tax, chartist share, N, T, seed) apply when you press **Reset with settings**.
- **Steps / frame** and **window** change the view while it plays.
- Panels: price vs fundamental value with the bid–ask spread; the live order book; mispricing; traded volume; and each trader type's share of trades and its share of activations skipped because of the tax.

Tip: run it once at τ = 0.6% and once at τ = 0.1% with the same seed to see the PH cut in the model. This illustrates the model's mechanism, not the PSE.

```python
viewer = live_ui(base)
```
